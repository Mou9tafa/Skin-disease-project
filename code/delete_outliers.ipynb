```python
import os
from PIL import Image
import numpy as np


base_dir = "/content/drive/MyDrive/data"

# إحصائيات عامة
total_outliers = 0
total_images = 0

# المرور على كل كلاس
for cls in os.listdir(base_dir):
    class_path = os.path.join(base_dir, cls)
    if not os.path.isdir(class_path):
        continue

    sizes = []
    images = os.listdir(class_path)

    # حساب أحجام الصور
    for img_name in images:
        img_path = os.path.join(class_path, img_name)
        try:
            with Image.open(img_path) as img:
                sizes.append(np.prod(img.size))  # الطول × العرض
        except:
            print(f"⚠️ ملف غير صالح: {img_name}")
            sizes.append(0)

    sizes = np.array(sizes)
    Q1 = np.percentile(sizes, 25)
    Q3 = np.percentile(sizes, 75)
    IQR = Q3 - Q1

    # تحديد الحدود الدنيا والعليا
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # تحديد الصور الشاذة
    outliers = [images[i] for i, s in enumerate(sizes) if s < lower_bound or s > upper_bound]

    # طباعة الإحصائيات
    print(f"\n📁 Class: {cls}")
    print(f"📸 عدد الصور: {len(images)}")
    print(f"🚫 عدد الصور الشاذة (سيتم حذفها): {len(outliers)} ({(len(outliers)/len(images))*100:.2f}%)")

    # حذف الصور الشاذة
    for img_name in outliers:
        img_path = os.path.join(class_path, img_name)
        try:
            os.remove(img_path)
        except:
            print(f"❌ فشل حذف الصورة: {img_name}")

    total_outliers += len(outliers)
    total_images += len(images)

print("\n============================")
print(f"📊 إجمالي عدد الصور: {total_images}")
print(f"🚮 الصور التي تم حذفها: {total_outliers}")
print(f"📉 النسبة الكلية: {(total_outliers/total_images)*100:.2f}%")
print("============================")




# 5️⃣ Summary Report
print("\n============================")
print(f"📊 Total images: {total_images}")
print(f"🚮 Total outliers: {total_outliers}")
print(f"📉 Outlier ratio: {(total_outliers/total_images)*100:.2f}%")

print("============================")


```
